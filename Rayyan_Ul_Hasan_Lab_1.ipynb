```python
class Flight:
    def __init__(self, id, departure, destination, price):
        self.id = id
        self.departure = departure
        self.destination = destination
        self.price = price


flights = [
    Flight(101, "Karachi", "Dubai", 45000),
    Flight(102, "Karachi", "Dubai", 40000),
    Flight(103, "Karachi", "Dubai", 35000),
    Flight(104, "Lahore", "Dubai", 50000),
    Flight(105, "Karachi", "Istanbul", 70000)
]

n = len(flights)

# Ask the user to enter departure city
departure = input("Enter departure city: ")

# Ask the user to enter destination city
destination = input("Enter destination city: ")

found = 0
booked = 0

# Search for flights that match departure and destination
print("\n===== AVAILABLE FLIGHTS =====")

for i in range(n):

    if (flights[i].departure == departure and
        flights[i].destination == destination):

        print("Flight ID:", flights[i].id,
              "| Price: Rs.", flights[i].price)

        found = 1


# IF no flights are found THEN
if found == 0:
    print("\nNo flights available")
else:

    # Ask user to choose a flight ID
    choice = int(input("\nEnter Flight ID to book: "))

    # Book the selected flight
    for i in range(n):

        if (flights[i].id == choice and
            flights[i].departure == departure and
            flights[i].destination == destination):

            booked = 1
            break

    if booked == 1:
        print("\nFlight booked successfully")
    else:
        print("\nInvalid Flight ID")
```
